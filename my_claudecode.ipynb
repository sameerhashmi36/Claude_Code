```python
from langchain_anthropic import ChatAnthropic
from dotenv import load_dotenv
load_dotenv(dotenv_path=".venv/.env")
import os

if os.getenv("ANTHROPIC_API_KEY") is None:
    raise ValueError("ANTHROPIC_API_KEY environment variable is not set.")
else:
    print("ANTHROPIC_API_KEY environment variable is set.")
```

```python
from langchain_core.tools import tool
import subprocess

@tool
def create_directory(directory_name: str):
    """This tool creates a directory with the given name.
    
    Args:
        directory_name (str): The name of the directory to create.
    
    """
    if not os.path.exists(directory_name):
        os.makedirs(directory_name)
        return f"Directory '{directory_name}' created successfully."

@tool
def create_file(file_name: str, code: str):
    """This tool creates a Python file with the given name and writes the provided code into it.
    
    Args:
        file_name (str): The name of the Python file to create.
        code (str): The code to write into the Python file.
    
    """
    with open(file_name, 'w') as f:
        f.write(code)

@tool
def install_package(package_name: str):
    """This tool installs a package using pip.
    
    Args:
        package_name (str): The name of the package to install.
    """

    subprocess.check_call(['pip', 'install', package_name])


from langgraph.prebuilt import create_react_agent

llm = ChatAnthropic(model="claude-haiku-4-5-20251001", anthropic_api_key=os.getenv("ANTHROPIC_API_KEY"))

tools = [create_directory, create_file, install_package]

agent = create_react_agent(model=llm, tools=tools)
```

```python
task = (
    "Create a directory named 'my_project', create a Python file named 'dataframe.py' inside it"
    "write code to create a pandas DataFrame with 3 columns and 5 rows. Also, install the pandas package if it's not already installed."
)
```

```python
response = agent.invoke({"messages": [{"role": "user", "content": task}]})
```
